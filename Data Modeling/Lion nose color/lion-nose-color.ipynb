# Lion Nose Color

Can you estimate a lion's age from a photograph of its nose?

Lions are born with pink noses that gradually darken with age, developing black freckles until they eventually turn completely black.

The dataset `lion-nose-color.csv` contains data on 91 lions from the Serengeti, each monitored by researchers since birth. Because their exact ages are known, researchers can analyze photographs to determine the percentage of black on their noses and assess how well nose coloration predicts age.

#1.Create a linear regression model to predict age from proportion_black

```python
def clear_axe():
  ax = plt.gca()
  ax.spines[['top','bottom','left','right']].set_visible(False)
  ax.tick_params(axis ='both', length=0)
```

```python
lion_nose = LinearModel()
lion_nose_x = df['proportion_black']
lion_nose_y = df['age']
lion_nose.fit(lion_nose_x, lion_nose_y)
lion_nose.predict(lion_nose_x)
plt.scatter(lion_nose_x, lion_nose_y)
lion_nose.plot_model(lion_nose_x.min(), lion_nose_x.max(), color="black")
plt.title("Age vs Proportion Black")
plt.xlabel("Proportion Black")
plt.ylabel("Age")
clear_axe()
plt.tight_layout()
plt.show()
lion_nose.print_model_info()
```

#2.Use the model to estimate a lion's age when its nose is 50% black.

```python
age_at_50_percent_black = lion_nose.predict(50)
print(f"Estimated age for a lion with 50% black nose: {age_at_50_percent_black:.2f} years")
```

#3.Can you determine a lion's age when its nose is 100% black? Why or why not?

```python
nose_100_percent_black = lion_nose.predict(100)
print(f"Estimated age for a lion with 100% black nose: {nose_100_percent_black:.2f} years")
```

#Why not:

It requires extrapolation beyond the data
In typical lion nose–color datasets, the percentage of black increases with age but does not usually reach exactly 100% within the observed age range. Estimating the age at 100% black would mean extending the trend beyond the available data, which is statistically unreliable.

The relationship may not stay linear
Even if a linear model fits the observed data reasonably well, there is no guarantee the relationship remains linear at older ages. Nose darkening may plateau before reaching 100%.


Biological variation

1.Lions age differently:

2.Genetics

3.Environment

4.Health


Some lions may never reach a fully black nose, even at old age.

Prediction vs. explanation
A linear regression model can describe trends within the data, but it is not strong evidence for making precise biological claims outside that range.

Conclusion:
A model might mathematically give you an age for 100% black nose color, but it would be an unsupported extrapolation, not a scientifically valid estimate.
