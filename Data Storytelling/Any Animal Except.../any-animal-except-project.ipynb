```python
import pandas as pd
import matplotlib.pyplot as plt

word_trends = pd.read_csv('animal-word-trends-menageries.csv')

def plot_menagerie_animal( animal ):
    plt.figure(figsize=(8, 3))
    trend = word_trends.query("word == @animal")
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'The menagerie period and word frequency for "{animal}"')
    plt.axvspan(1750, 1835, color='magenta', label='Menagerie Period', alpha=0.2)
```

# British Menageries and the Circulation of Animal Words in English

## Abstract

This report examines the relationship between the rise of British menageries in the late eighteenth and early nineteenth centuries and changes in the frequency of animal-related words in English print culture. Using historical word-frequency data and focusing on a defined "menagerie period" (c. 1800–1835), the analysis compares animals commonly exhibited in British menageries with animals that were rare, difficult to keep, or absent from such collections. The findings suggest that public exhibition played a significant role in shaping which animal names entered more frequent linguistic circulation, highlighting menageries as an important cultural mediator between imperial exploration, popular entertainment, and language use.

## 1. Introduction

During the late eighteenth and early nineteenth centuries, British menageries—both private and public—expanded rapidly. Fueled by imperial trade, exploration, and colonial networks, these collections displayed exotic animals to metropolitan audiences who would otherwise never encounter them. While historians have examined menageries as sites of spectacle and knowledge production, their influence on everyday language has received less attention.

This report investigates whether the presence of animals in British menageries corresponded with measurable increases in the frequency of animal words in English texts. The central research question is: **Did exhibition in menageries contribute to increased usage of specific animal names in print, and if so, how did this effect vary across different types of animals?**

## 2. Data and Methodology

The analysis draws on a historical word-frequency dataset tracking the relative usage of animal names in English-language texts over time. A specific window—the menagerie period (approximately 1800–1835)—is highlighted to capture the height of menagerie popularity in Britain.

Animals were grouped into two main categories:

1. **Menagerie animals**: Exotic species that were commonly and successfully exhibited (e.g., tiger, hyena, tapir, toucan, macaw, ostrich).
2. **Difficult or rarely exhibited animals**: Species that were hard to transport, keep alive, or display for long periods (e.g., hummingbird, chimpanzee, meerkat, penguin).

Trends in word frequency for each animal were visually inspected and compared across these groups, with particular attention to changes during and after the menagerie period.

## 3. Results

### 3.1 Common Menagerie Animals

Animal words associated with frequently exhibited menagerie species show a clear and shared pattern. For many of these animals, word frequency rises noticeably during the menagerie period. In some cases, this increase begins slightly before 1800 and peaks during the early nineteenth century.

This pattern suggests that repeated public exposure—through advertisements, newspaper reports, guidebooks, and popular writing—helped normalize these animal names in English discourse. Animals such as tigers or ostriches became familiar reference points rather than purely exotic curiosities.

### 3.2 Difficult or Rarely Exhibited Animals

In contrast, animals that were difficult to keep or rarely displayed show weaker or delayed increases in word frequency. Some exhibit little change during the menagerie period, while others only rise in frequency later in the nineteenth century, often coinciding with advances in zoology, travel writing, or scientific classification.

The comparison indicates that mere imperial access to animals was not sufficient to drive word usage. Sustained public visibility and repeat encounters appear to have been key factors.

### 3.3 Comparative Patterns

The contrast between the two groups highlights a divergence in linguistic impact. Animals that could be reliably exhibited generated a stronger and earlier linguistic footprint than those that remained biologically or logistically challenging for menageries.

## 4. Explanatory Mechanism

The observed differences can be explained through the cultural function of menageries. Menageries acted as translation devices: they transformed distant, unfamiliar animals into recognizable cultural objects. Exhibition encouraged discussion, description, and comparison, all of which increased the likelihood that an animal’s name would appear in print.

Animals absent from menageries lacked this amplification mechanism. Even if they were known through exploration or science, they did not benefit from the same level of popular exposure.

## 5. Case Study: Australian Animals and the Koala

Australian animals provide a useful test of this explanation. Many were geographically remote and biologically specialized. Among them, the koala stands out as particularly difficult to keep in captivity due to its highly specific diet.

Consistent with this constraint, the word *koala* shows relatively limited growth in frequency during the menagerie period, with more substantial increases occurring later, when scientific and popular writing about Australia expanded. This supports the hypothesis that successful exhibition, rather than mere discovery, was crucial for early increases in word usage.

## 6. Discussion

These findings suggest that linguistic change can be shaped by cultural institutions as much as by exploration or scientific discovery. Menageries did not simply reflect existing interest in exotic animals; they actively structured which animals became linguistically salient to the British public.

This perspective reframes menageries as part of the infrastructure of language change, linking entertainment, empire, and print culture.

## 7. Conclusion

The analysis demonstrates a clear association between British menageries and increased word frequency for certain animal names during the early nineteenth century. Animals that were commonly exhibited show stronger and earlier rises in usage than those that were difficult or impossible to display.

British menageries thus played a significant role in shaping how the animal world entered everyday English, leaving traces not only in visual culture and science but also in the historical record of language itself.
