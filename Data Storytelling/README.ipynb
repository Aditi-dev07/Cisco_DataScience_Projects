# 📊 Master Portfolio Notebook: Language, Nature & Data Science
This notebook serves as a **unified narrative overview** of seven data science projects exploring how history, culture, environment, and climate leave measurable traces in language.

Each section summarizes a standalone project while contributing to a cohesive analytical portfolio.

## 🐅 Project 1: Any Animal Except…

Explores how the rise of British menageries (1750–1835) influenced the frequency of exotic animal words in English. Using time-series word-frequency data, the project shows how cultural exposure precedes linguistic adoption.

## 🪸 Project 2: Coral Conundrum

Investigates the sudden decline in coral reef fish terminology around 1998. By combining linguistic data with coral bleaching records and scuba certification trends, the project links environmental stress to changes in public discourse.

## 🐾 Project 3: Plotting Animal Words

Focuses on visualization engineering by building reusable, consistent plotting functions for time-series language analysis. Emphasizes clarity, modularity, and interpretability over complex modeling.

## 🧬 Project 4: Primate Research

Analyzes the rise of primate-related terminology over the 20th century, aligning linguistic trends with advances in evolutionary biology and scientific research.

## 🐦 Project 5: Silent Spring

Examines how Rachel Carson’s *Silent Spring* (1962) reshaped environmental language. Event-based time-series analysis reveals shifts in how nature and ecology entered public discourse.

## 🦗 Project 6: Cicadas Are Coming

Tracks spikes in cicada-related language corresponding to 13- and 17-year emergence cycles. Demonstrates how rare but predictable natural events produce intense yet fleeting public attention.

## 🌊 Project 7: Warm Waters off Peru

Combines sea surface temperature anomalies with linguistic trends to detect the impact of El Niño events. Shows how large-scale climate disruptions influence both ecosystems and language.

## 🧠 Core Skills Demonstrated Across Projects
- Time-series analysis
- Event-based and exploratory data analysis
- Data visualization and storytelling
- Multi-dataset integration
- Historical, cultural, and environmental reasoning

## 🚀 Portfolio Takeaway
Together, these projects demonstrate an ability to move beyond metrics—connecting data with real-world phenomena and translating complex patterns into meaningful insights.
