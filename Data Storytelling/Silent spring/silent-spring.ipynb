# Silent Spring

When the pesticide DDT was made available to the public in 1944, it was seen as a miracle chemical in the battle against insect pests. How did the worldwide availability of DDT coincide with word trends for insect words?

```python
# FOR GOOGLE COLAB ONLY.
# Uncomment and run the code below. A dialog will appear to upload files.
# Upload 'animal-word-trends-silent-spring.csv' and 'sparrowhawk-eggshell-data.csv'.

from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('animal-word-trends-silent-spring.csv')
df.head()
```

```python
def plot_insect_word_trend( animal ):
    trend = df.query('word == @animal')
    plt.figure(figsize=(5, 3))
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'Word usage for "{animal}" and the year DDT was made available')
    plt.axvline(1944, color='orange', ls='--')
```

Explore word trends for insects relative to 1944.

```python
plot_insect_word_trend('flea')
# try flea, mosquito, fly, bedbug, louse
```

```python
plot_insect_word_trend('mosquito')
```

```python
plot_insect_word_trend('fly')
plot_insect_word_trend('bedbug')
```

```python
plot_insect_word_trend('louse')
```

#### Impact of DDT

Early marketing for DDT claimed it only killed insect pests and did not affect other animals. The following python script might help shed light on this claim.

```python
egg = pd.read_csv('sparrowhawk-eggshell-data.csv')
plt.figure(figsize=(5, 3))
plt.plot(egg['year'],egg['avg_thickness'], color='C5')
plt.xlabel('Year')
plt.ylabel('Thickness Index')
plt.title('Average Eggshell Thickness')
plt.axvline( 1944, color='magenta', ls='--', label='DDT made available to public' )
plt.show()
```

#### Rachel Carson

In 1962, Rachel Carson published the book Silent Spring, which showed that the toxins killing insects were also harming the rest of the food web. This book began to raise public awareness of DDT's toxicity, which eventually lead to bans on DDT.

In particular, birds of prey such as peregrine falcons, and songbirds such as robins, were highly affected by widespread use of DDT.

Use the function below to explore the word usage levels for these birds relative to the 1962 book.


```python
def plot_bird_word_trend( animal ):
    trend = df.query('word == @animal')
    plt.figure(figsize=(5, 3))
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'Word usage for "{animal}" and the year Silent Spring published')
    plt.axvline(1962, color='orange', ls='--')
```

```python
plot_bird_word_trend('robin')
plot_bird_word_trend('hawk')
plot_bird_word_trend('peregrine')
plot_bird_word_trend('osprey')
plot_bird_word_trend('sparrowhawk')
plot_bird_word_trend('condor')
# try robin, hawk, peregrine, osprey, sparrowhawk, condor
```

The Impact of DDT on Cultural Discourse Regarding Birds and Insects
Overview This analysis examines the frequency of bird and insect terminology in written texts from 1900 to the present. The data reveals a significant divergence in how these two groups were discussed following the mid-20th century, specifically correlating with the public introduction of the pesticide DDT in 1944.

**1.Diverging Trends: A Shift in Focus**
The year 1944 serves as a clear pivot point in the data, marking the beginning of opposing trends for birds and insects.


The Decline of the Insect: Prior to 1944, insects were a frequent topic, with related words (e.g., fly, mosquito, flea) appearing with a high average frequency of 34.6. However, in the post-1944 era, this prevalence dropped sharply to an average of 23.9. The data establishes a consistent downward trend (slope: -0.15), suggesting that as chemical controls became available, insects faded from the forefront of public concern.

The Rise of the Bird: Conversely, bird terminology (e.g., hawk, robin, eagle) started from a lower baseline of 14.5 pre-1944. in the years following, bird references have seen a steady resurgence, rising to an average of 16.3 with a positive growth trend (slope: +0.16).


**2.Drivers of Change: The DDT Factor**


The introduction of DDT provides a compelling historical context for these statistical shifts.

Conquering the Pest: The drop in insect word usage likely reflects the "success" of the war on pests. As DDT and other pesticides effectively reduced insect nuisances in daily life, the cultural necessity to discuss them diminished.

The Environmental Awakening: The rise in bird discourse is closely tied to the environmental movement that emerged in response to pesticide use, epitomized by Rachel Carson’s Silent Spring (1962). The very chemicals used to silence insects had unintended, devastating effects on bird populations—specifically causing eggshell thinning in raptors like the Sparrowhawk.


 **3.Ecological Evidence in the Text**


The analysis cross-referenced Sparrowhawk eggshell thickness data with the word frequency of "sparrowhawk."

Observations: While the direct statistical correlation is weak (-0.11), a visual inspection of the timeline tells a clearer story. As physical eggshell thickness plummeted due to DDT contamination, the cultural conversation (word frequency) around these birds eventually trended upward. This lag suggests that the scientific crisis fueled a delayed but sustained increase in public awareness and conservation discourse.

**Conclusion:**

The linguistic data mirrors the ecological history of the 20th century. The introduction of DDT successfully pushed insects to the margins of our language, but the resulting ecological damage brought birds back to the center of it. The trends document a shift from a society focused on eradicating pests to one increasingly concerned with preserving wildlife.
