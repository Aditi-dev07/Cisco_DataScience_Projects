```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', 10)

word_trends = pd.read_csv('animal-word-trends-peru.csv')

def plot_peru_animal( animal ):
	plt.figure(figsize=(8, 3))
	trend = word_trends.query("word == @animal")
	plt.plot(trend['year'], trend['frequency'], label=animal)

	for year in [1965, 1973, 1983, 1987]:
		plt.axvline( year, label='El Niño event', color='magenta', ls='--' )

	plt.ylabel('Frequency per million')
	plt.title(f'El Niño events and word frequency for "{animal}"')
	plt.xlim(1960, 1990)
	plt.ylim(0, None)
	plt.gca().spines[['top', 'right']].set_visible(False)
```

## El Niño Events and Fish-Related Word Usage in Peru: A Research Report

### 1. Introduction
El Niño is a natural climate phenomenon characterized by the warming of ocean waters in the equatorial Pacific, which has profound impacts on global weather patterns. Centuries ago, Peruvian fishing communities observed these warm waters and coined the term 'El Niño' to describe years when these conditions adversely affected fishing productivity, highlighting its long-standing historical significance and impact on their livelihoods.

This study aims to investigate the relationship between these El Niño events and word usage levels for various economically important fish species in Peru. By analyzing word trends, we seek to understand how these significant climatic events influenced public discourse and perception regarding different fish populations, potentially reflecting changes in their abundance, market value, or societal concern.

### 2. Methodology
The methodology involved analyzing two primary datasets: `animal-word-trends-peru.csv` and `anchoveta-caught-per-year.csv`. Word frequency data for five economically important fish species ('sardine', 'mackerel', 'bonito', 'anchoveta', and 'hake') was plotted over time, with major El Niño years (1965, 1973, 1983, 1987) marked to observe any correlations. Similarly, annual anchoveta catch data (in megatonnes) was plotted, with El Niño years also indicated. Visual inspection and qualitative analysis were employed to identify patterns, trends, and relationships between El Niño events and both word usage frequency and fish catch sizes.

### 3. Results
#### Fish Word Usage and El Niño
Observations from plotting word frequencies against El Niño years revealed distinct patterns for different fish species. While 'sardine', 'mackerel', 'bonito', and 'hake' showed less consistent or no strong correlation with El Niño events, 'anchoveta' demonstrated a clear and significant pattern.

**Detailed Observations for 'Anchoveta':**
The word 'anchoveta' shows a noticeable increase in frequency of usage around the major El Niño years (1965, 1973, 1983, 1987). For instance, there's a clear spike in word usage in the year of, or immediately following, these events. This pattern is more pronounced than for other fish words, suggesting a unique relationship between anchoveta discussions and El Niño phenomena.

#### Anchoveta Catch Data and El Niño
The plot of 'Megatonnes of Anchoveta Caught per Year' clearly shows a significant decline in anchoveta catch during or immediately following the El Niño events marked by the magenta vertical lines (1965, 1973, 1983, 1987). For instance, there are noticeable drops in catch around 1965, 1973, and a very dramatic decrease in 1983 and 1987.

This trend directly refutes the **Abundance hypothesis**, which suggested anchoveta might thrive in warm waters and become more abundant during El Niño years. Instead, the data indicates the opposite: anchoveta populations or catchability are negatively impacted during these events.

The observations strongly support the **Scarcity hypothesis**. The sharp declines in caught anchoveta suggest that their populations indeed suffer in warm water conditions brought by El Niño. This scarcity would naturally lead the fishing community to be concerned and discuss anchoveta more, aligning with the increased word frequency observed previously.

The **Exception hypothesis**, which proposed anchoveta might be an exception to El Niño's impact, is also refuted. The catch data clearly demonstrates that anchoveta fisheries are significantly affected, experiencing substantial reductions in yield during El Niño years.

### 4. Discussion
This analysis reveals a compelling inverse relationship between the frequency of the word "anchoveta" in public discourse and the actual catch sizes of anchoveta during El Niño events. For each major El Niño event, the discussion around "anchoveta" intensifies, while the amount of anchoveta caught significantly decreases.

The timing of these changes provides further insight. The increase in "anchoveta" word usage often begins to rise well before the El Niño event itself, suggesting a growing awareness or concern within the fishing community as the climatic phenomenon approaches. This could be driven by early signs of warming waters or anecdotal reports of changes in fish behavior. The reduced catch size, however, is more directly tied to the year of the El Niño event, indicating a more immediate impact on fishing productivity.

This inverse relationship is most plausibly explained by the "Scarcity hypothesis." As El Niño conditions lead to warmer waters and a subsequent decline in anchoveta populations or their accessibility, the crucial role of anchoveta in the Peruvian economy and diet would prompt heightened public and professional discussion. This discussion would encompass concerns about food security, economic impact on fishing communities, and potential conservation efforts. The increased discourse around a diminishing resource reflects the significant social and economic implications of such a decline.

### 5. Conclusion
This report has demonstrated a clear inverse relationship between El Niño events and the usage frequency of the word "anchoveta" and its catch sizes. El Niño years consistently correlate with an increase in "anchoveta" discussions and a sharp decline in anchoveta catches. This pattern refutes the abundance and exception hypotheses, strongly supporting the scarcity hypothesis: as anchoveta become scarcer due to El Niño's impact, public and professional concern intensifies, leading to increased word usage. The timing suggests that elevated discourse often precedes the full impact on catch, highlighting the fishing community's proactive awareness of environmental changes and their potential economic and social consequences. This study underscores the profound influence of natural climatic events on both ecological systems and human societal discourse.
