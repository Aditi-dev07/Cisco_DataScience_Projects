# Primate Research

Can books change cultures?

The 1975 publication of the book Animal Liberation by Peter Singer is often credited with sparking an animal rights movement. Did it? Let's look at word usage data to explore this claim? We will focus on primates used in biomedical research.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('animal-word-trends-primate-research.csv')
df.head()
```

```python
def plot_word_trend( animal ):
    trend = df.query('word == @animal')
    plt.figure(figsize=(5, 3))
    plt.plot(trend['year'], trend['frequency'], label=animal)
    plt.ylabel('Frequency per million')
    plt.title(f'Word usage for "{animal}" and year Animal Liberation book published')
    plt.axvline(1975, color='orange', ls='--')
```

Explore word trends for primates used in research.

```python
plot_word_trend('primate')
plot_word_trend('marmoset')
plot_word_trend('baboon')
plot_word_trend('rhesus')
# try primate, marmoset, baboon, rhesus
```

### How animal research changed

Let's look at another dataset to see what was happening with the research.

```python
counts = pd.read_csv('number-of-research-animals-per-year.csv')
counts.head(2)
```

```python
def plot_animal_research( animal ):
	plt.figure(figsize=(5, 3))
	plt.plot(counts['year'], counts[animal])
	plt.axvline(1975, label='Animal Liberation book', color='orange', ls='--')
	plt.ylabel('Number of animals')
	plt.title(f'Number of {animal} used in research per year')
	plt.legend()
	plt.show()

plot_animal_research('primates')
```

1.Word Usage Trends for Primates

The word usage data shows a significant and sustained increase in the frequency of primate-related terms following the 1975 publication of Animal Liberation.

Broad Increase: All tracked terms—"primate," "rhesus," "baboon," and "marmoset"—saw their usage frequencies rise sharply after 1975.

Term Frequency: The term "primate" saw the most dramatic growth, more than doubling from an average frequency of approximately 1.85 before 1975 to over 4.66 in the years following.

Specificity: Other specific terms like "rhesus" and "baboon" also nearly doubled in frequency, suggesting a heightened visibility and discussion of these animals in written culture.

2.Trends in the Number of Primates Used in Research
Contrary to what one might expect if a movement successfully reduced animal testing across the board, the number of primates used in research has actually increased since 1975.

Primate Usage Growth: In 1975, approximately 36,202 primates were used in research. By 2019, that number had climbed to 68,257, representing an increase of over 61%.

Contrast with Other Animals: This trend is unique to primates. During the same period (1973–2019), the use of other common research animals dropped significantly:

Dogs: -70%

Cats: -72%

Rabbits: -68%

Hamsters: -78%

3.Potential Explanations for Word Usage Trends
The discrepancy between the increased "talk" about primates and the increased "use" of primates can be explained by several factors:

Increased Ethical Scrutiny: The rise in word frequency likely reflects the growing ethical debate sparked by the animal rights movement. As primates are more biologically similar to humans, they became a central focus of advocacy, leading to more literature, legislation, and public discourse.

Scientific Specialization: While the movement may have pushed research away from cats and dogs, primates remained (or became) indispensable for specific high-stakes medical research (e.g., neuroscience, immunology, and vaccine development), leading to more scientific papers mentioning them.

Linguistic Shift: There may have been a shift toward using more general and scientifically inclusive terms like "primate" rather than just naming specific species.

4.Conclusions on the Claim:

Did Animal Liberation Spark a Movement?
The data provides a nuanced answer to whether Peter Singer’s book sparked a successful animal rights movement:

Success for Most Species: There is strong evidence that a movement occurred and was effective for many animals. The massive, 70%+ declines in the use of dogs, cats, and rabbits suggest a cultural and regulatory shift that moved research away from these species.

The "Primate Exception": For primates, the claim is more complex. While the movement clearly increased public awareness (as seen in the word usage trends), it did not result in a decrease in the number of primates used. Instead, primates appear to have become more central to the research industry even as other animals were phased out.

Overall Conclusion: The book Animal Liberation likely did spark a movement that fundamentally changed how society views animal research, leading to a drastic reduction in the use of many domestic animals. However, this movement has been less successful in reducing the scientific reliance on primates.
